```python
# -*- coding: utf-8 -*-
"""RL_Trading_BTC_V1_5.ipynb

Automatically generated by Colab.

Original file is located at
    https://colab.research.google.com/drive/10dh_dtqVuPhqEJG5ztO5pknflStjOEcs

# Prototype V1.5 — Agent RL de trading BTC/USDT (Binance + Walk-Forward)

Version consolidée qui reprend le V1 (yfinance, split unique) et ajoute tout ce qu'on a construit ensemble :
- Données Binance (plus d'historique que yfinance, couvre plusieurs régimes bull/bear)
- Walk-forward rolling window multi-folds (au lieu d'un seul split)
- `EvalCallback` (garde le meilleur checkpoint sur validation, évite l'overfitting observé en V1)
- Config retenue après tests V1 : SL fixe 1%, ratio SL/TP = 1:2
- Diagnostics : plafonnement de l'entraînement, taux de trades, Sharpe (le Sortino n'est pas fiable ici, expliqué en section 8)

**Budget de temps réaliste : avec `N_FOLDS=2` et `TIMESTEPS=300_000`, compte 40-60 minutes au total sur Colab gratuit.** Augmente ces valeurs seulement si tu as le temps devant toi.

Exécute les cellules dans l'ordre, avec Shift+Entrée.


## 1. SYNC + Installation des librairies
"""

# ============================================================
# SYNC GITHUB — à coller en TOUTE PREMIÈRE cellule de chaque
# session Colab, avant tout le reste du notebook.
#
# Récupère la dernière version du code, et fournit REPO_PATH
# (réutilisé par le bloc de logging Google Sheets en fin de notebook).
#
# Pré-requis (une fois, pour chaque personne) :
#   1. Créer un Personal Access Token GitHub (fine-grained, scope
#      "Contents: read/write" limité à ce repo).
#   2. Dans Colab : icône clé "Secrets" (panneau de gauche) > ajouter
#      un secret nommé GITHUB_TOKEN, coller le token, activer l'accès
#      notebook.
# ============================================================

from google.colab import userdata
import subprocess
import os

# --- À adapter une seule fois, identique pour toi et ton collègue ---
GITHUB_REPO = "github.com/votre-org/votre-repo.git"  # sans le https://
REPO_PATH = "/content/repo"
# ----------------------------------------------------------------

GITHUB_TOKEN = userdata.get('GITHUB_TOKEN')


def _run(cmd, cwd=None):
    result = subprocess.run(cmd, shell=True, cwd=cwd, capture_output=True, text=True)
    if result.returncode != 0 and result.stderr.strip():
        print(f"⚠️  {cmd}\n{result.stderr.strip()}")
    return result.stdout.strip()


if not os.path.exists(REPO_PATH):
    print("Clonage du dépôt...")
    _run(f"git clone https://{GITHUB_TOKEN}@{GITHUB_REPO} {REPO_PATH}")
else:
    print("Dépôt déjà présent, récupération des derniers changements (git pull)...")
    _run("git pull", cwd=REPO_PATH)

_commit = _run("git rev-parse --short HEAD", cwd=REPO_PATH) or "unknown"
print(f"✅ Code synchronisé — commit courant : {_commit}")


def push_to_github(commit_message):
    """À appeler manuellement (dans une cellule séparée) si tu as modifié
    le notebook directement dans Colab et veux renvoyer les changements
    sur GitHub. Pas nécessaire si tu édites le code en local et push
    depuis ta machine."""
    _run("git add -A", cwd=REPO_PATH)
    print(_run(f'git commit -m "{commit_message}"', cwd=REPO_PATH))
    _run(f"git push https://{GITHUB_TOKEN}@{GITHUB_REPO}", cwd=REPO_PATH)
    print("✅ Poussé sur GitHub.")


!pip install -q gymnasium stable-baselines3 pandas numpy matplotlib requests

import numpy as np
import pandas as pd
import requests
import matplotlib.pyplot as plt
import gymnasium as gym
from gymnasium import spaces
from stable_baselines3 import PPO
from stable_baselines3.common.monitor import Monitor
from stable_baselines3.common.callbacks import EvalCallback
from stable_baselines3.common.env_checker import check_env

print("Librairies chargées.")

"""## 2. Téléchargement des données — Binance (historique riche, gratuit, sans clé API)

4 ans d'historique horaire, couvrant le bull 2021, le bear 2022, et la remontée qui suit — plusieurs régimes indépendants, contrairement aux 2 ans limités de yfinance utilisés en V1.
"""

def get_binance_klines(symbol="BTCUSDT", interval="1h", start_str="2020-06-01", end_str="2024-06-01"):
    url = "https://data-api.binance.vision/api/v3/klines"
    start_ts = int(pd.Timestamp(start_str).timestamp() * 1000)
    end_ts = int(pd.Timestamp(end_str).timestamp() * 1000)
    all_data = []
    while start_ts < end_ts:
        params = {"symbol": symbol, "interval": interval, "startTime": start_ts, "endTime": end_ts, "limit": 1000}
        r = requests.get(url, params=params).json()
        if not r or isinstance(r, dict):
            print("Erreur API:", r)
            break
        all_data += r
        start_ts = r[-1][0] + 1
    cols = ["open_time","open","high","low","close","volume","close_time","qav","trades","tbb","tbq","ignore"]
    df = pd.DataFrame(all_data, columns=cols)
    df = df[["open","high","low","close","volume"]].astype(float)
    return df.reset_index(drop=True)

df_full = get_binance_klines(start_str="2020-06-01", end_str="2024-06-01")
print(f"{len(df_full)} bougies téléchargées.")
df_full.tail()

"""## 3. Environnement de trading (Gymnasium)

Identique au V1 — inchangé, c'est cette partie qui a déjà été validée (check_env OK, logique de récompense testée).

**État :** fenêtre glissante de 20 bougies (rendement log du close/high/low, volume normalisé, volatilité réalisée) + position actuelle.
**Actions :** 0 = rien, 1 = long, 2 = short.
**Récompense (4 termes) :** R-multiple net de frais à la clôture, signal intermédiaire faible en position, coût de portage, garde-fou de drawdown.
"""

class BTCTradingEnv(gym.Env):
    metadata = {"render_modes": []}

    def __init__(self, df, window_size=20, sl_pct=0.01, tp_mult=2.0, fee_pct=0.0004, max_drawdown=0.2):
        super().__init__()
        self.df = df.reset_index(drop=True)
        self.window_size = window_size
        self.sl_pct = sl_pct
        self.tp_pct = sl_pct * tp_mult
        self.fee_pct = fee_pct
        self.max_drawdown = max_drawdown

        log_ret = np.log(self.df["close"] / self.df["close"].shift(1)).fillna(0)
        high_ret = np.log(self.df["high"] / self.df["close"].shift(1)).fillna(0)
        low_ret = np.log(self.df["low"] / self.df["close"].shift(1)).fillna(0)
        vol_ma = self.df["volume"].rolling(20).mean().bfill()
        vol_norm = (self.df["volume"] / vol_ma).fillna(1.0)
        realized_vol = log_ret.rolling(window_size).std().bfill()

        self.features = np.column_stack([log_ret, high_ret, low_ret, vol_norm, realized_vol]).astype(np.float32)
        self.n_features = self.features.shape[1]

        self.action_space = spaces.Discrete(3)
        self.observation_space = spaces.Box(low=-np.inf, high=np.inf, shape=(window_size * self.n_features + 1,), dtype=np.float32)

    def _get_obs(self):
        window = self.features[self.current_step - self.window_size: self.current_step].flatten()
        return np.concatenate([window, [self.position]]).astype(np.float32)

    def reset(self, seed=None, options=None):
        super().reset(seed=seed)
        self.current_step = self.window_size
        self.position = 0
        self.entry_price = None
        self.equity = 1.0
        self.peak_equity = 1.0
        self.trade_log = []
        return self._get_obs(), {}

    def step(self, action):
        price = self.df["close"].iloc[self.current_step]
        reward = 0.0
        terminated = False

        if self.position == 0:
            if action == 1:
                self.position = 1; self.entry_price = price
            elif action == 2:
                self.position = -1; self.entry_price = price
        else:
            direction = self.position
            move_pct = (price - self.entry_price) / self.entry_price * direction
            hit_tp = move_pct >= self.tp_pct
            hit_sl = move_pct <= -self.sl_pct

            if hit_tp or hit_sl:
                realized_pnl = self.tp_pct if hit_tp else -self.sl_pct
                fees = self.fee_pct * 2
                risk_unit = self.sl_pct
                reward += (realized_pnl - fees) / risk_unit
                self.equity *= (1 + realized_pnl - fees)
                self.trade_log.append(realized_pnl - fees)
                self.position = 0; self.entry_price = None
            else:
                reward += move_pct * 0.01
                reward -= 0.00002

            self.peak_equity = max(self.peak_equity, self.equity)
            drawdown = (self.peak_equity - self.equity) / self.peak_equity
            if drawdown > self.max_drawdown:
                reward -= 1.0

        self.current_step += 1
        if self.current_step >= len(self.df) - 1:
            terminated = True

        return self._get_obs(), reward, terminated, False, {"equity": self.equity}

"""## 4. Découpage walk-forward — Rolling Window multi-folds

Contrairement au split unique du V1, on génère ici plusieurs fenêtres (folds) qui glissent dans le temps — chaque fold a son propre train/val/test, l'historique le plus ancien est progressivement écarté. C'est ce qui permet de vérifier la robustesse de la config sur plusieurs périodes indépendantes, pas juste une seule.
"""

def generate_rolling_folds(n_total, n_folds=2, window_frac=0.55, train_frac=0.70, val_frac=0.15, embargo=50):
    W = int(n_total * window_frac)
    S = (n_total - W) // (n_folds - 1) if n_folds > 1 else 0
    folds = []
    for i in range(n_folds):
        start = i * S
        end = min(start + W, n_total)
        start = end - W
        train_end = start + int(W * train_frac)
        val_end = train_end + int(W * val_frac)
        folds.append({
            "train": (start, train_end - embargo),
            "val": (train_end, val_end - embargo),
            "test": (val_end, end),
        })
    return folds

N_FOLDS = 4  # augmente à 4 seulement si tu as le temps (voir budget de temps en intro)
EXPERIMENT_CONFIG = {"n_folds": N_FOLDS, "window_frac": 0.55, "train_frac": 0.70, "val_frac": 0.15, "embargo": 50}
folds = generate_rolling_folds(len(df_full), **EXPERIMENT_CONFIG)

for i, f in enumerate(folds):
    print(f"Fold {i+1}: train={f['train'][1]-f['train'][0]} bars | "
          f"val={f['val'][1]-f['val'][0]} bars | test={f['test'][1]-f['test'][0]} bars")

"""## 5. Config retenue (issue des tests V1)

`sl_pct=0.01, tp_mult=2.0` — c'est la config qui a donné les meilleurs résultats en V1 sur yfinance (bull, bear, et mix), confirmée cohérente sur 3 seeds différents (variance observée : équity finale entre +13% et +55% selon le seed, moyenne ~+38%).

`TIMESTEPS = 300_000` par fold (réduit vs les 500k du V1 pour tenir le budget de temps avec plusieurs folds — le plafonnement observé en V1 arrivait déjà vers 300-400k timesteps, donc c'est un compromis raisonnable, pas une coupe arbitraire).
"""

SL_PCT = 0.01
TP_MULT = 2.0
TIMESTEPS = 200_000
SEED = 121
MAX_DRAWDOWN = 0.2
FEE_PCT = 0.0004

PIPELINE_TYPE = "rl_ppo"
PIPELINE_CONFIG = {
    "sl_pct": SL_PCT,
    "tp_mult": TP_MULT,
    "timesteps": TIMESTEPS,
    "seed": SEED,
    "max_drawdown": MAX_DRAWDOWN,
    "fee_pct": FEE_PCT,
}

"""## 6. Entraînement — boucle sur tous les folds

Pour chaque fold : entraîne avec `EvalCallback` (garde le meilleur checkpoint selon la validation de CE fold, évite l'overfitting comme identifié en V1), puis évalue sur la validation de ce fold.
"""

results = []
fold_equity_curves = []
fold_training_rewards = []

for fold_idx, f in enumerate(folds):
    print(f"\n=== Fold {fold_idx+1}/{len(folds)} ===")
    df_train_f = df_full.iloc[f["train"][0]:f["train"][1]].reset_index(drop=True)
    df_val_f   = df_full.iloc[f["val"][0]:f["val"][1]].reset_index(drop=True)

    env_train = Monitor(BTCTradingEnv(df_train_f, sl_pct=SL_PCT, tp_mult=TP_MULT))
    env_val_cb = Monitor(BTCTradingEnv(df_val_f, sl_pct=SL_PCT, tp_mult=TP_MULT))

    eval_cb = EvalCallback(
        env_val_cb,
        best_model_save_path=f"./best_model_fold{fold_idx}/",
        eval_freq=10_000,
        n_eval_episodes=1,
        deterministic=True,
    )

    model = PPO("MlpPolicy", env_train, verbose=0, seed=SEED)
    model.learn(total_timesteps=TIMESTEPS, callback=eval_cb)
    fold_training_rewards.append(env_train.get_episode_rewards())

    model = PPO.load(f"./best_model_fold{fold_idx}/best_model.zip")

    # Évaluation sur la validation de ce fold
    env_val = BTCTradingEnv(df_val_f, sl_pct=SL_PCT, tp_mult=TP_MULT)
    obs, _ = env_val.reset()
    equity_curve = [1.0]
    done = False
    while not done:
        action, _ = model.predict(obs, deterministic=True)
        obs, reward, terminated, truncated, info = env_val.step(action)
        equity_curve.append(info["equity"])
        done = terminated or truncated

    n_trades = len(env_val.trade_log)
    win_rate = np.mean([t > 0 for t in env_val.trade_log]) if n_trades > 0 else 0
    trades_per_1000 = (n_trades / len(df_val_f)) * 1000

    returns_arr = np.diff(np.log(equity_curve))
    sharpe = returns_arr.mean() / returns_arr.std() * np.sqrt(24 * 365) if returns_arr.std() > 0 else np.nan

    equity_arr = np.array(equity_curve)
    running_max = np.maximum.accumulate(equity_arr)
    max_dd = ((running_max - equity_arr) / running_max).max()

    fold_equity_curves.append(equity_curve)
    results.append({
        "fold": fold_idx + 1,
        "n_trades": n_trades,
        "trades_per_1000": round(trades_per_1000, 1),
        "win_rate_%": round(win_rate * 100, 1),
        "equity_final": round(equity_curve[-1], 3),
        "max_drawdown_%": round(max_dd * 100, 1),
        "sharpe_annualized": round(sharpe, 2),
    })
    print(f"Fold {fold_idx+1} — win_rate={win_rate*100:.1f}% | equity={equity_curve[-1]:.3f} | "
          f"drawdown={max_dd*100:.1f}% | Sharpe={sharpe:.2f}")

results_df = pd.DataFrame(results)
print("\n=== Résumé walk-forward (tous les folds) ===")
print(results_df.to_string(index=False))

# ============================================
# CELLULE 6.5 — Évaluation finale sur TEST (données jamais vues, un seul passage)
# ============================================
print("=== Évaluation finale sur TEST — données jamais utilisées jusqu'ici ===\n")

test_results = []
for fold_idx, f in enumerate(folds):
    df_test_f = df_full.iloc[f["test"][0]:f["test"][1]].reset_index(drop=True)
    model = PPO.load(f"./best_model_fold{fold_idx}/best_model.zip")

    env_test = BTCTradingEnv(df_test_f, sl_pct=SL_PCT, tp_mult=TP_MULT)
    obs, _ = env_test.reset()
    equity_curve = [1.0]
    done = False
    while not done:
        action, _ = model.predict(obs, deterministic=True)
        obs, reward, terminated, truncated, info = env_test.step(action)
        equity_curve.append(info["equity"])
        done = terminated or truncated

    n_trades = len(env_test.trade_log)
    win_rate = np.mean([t > 0 for t in env_test.trade_log]) if n_trades > 0 else 0
    equity_arr = np.array(equity_curve)
    running_max = np.maximum.accumulate(equity_arr)
    max_dd = ((running_max - equity_arr) / running_max).max()
    returns_arr = np.diff(np.log(equity_curve))
    sharpe = returns_arr.mean() / returns_arr.std() * np.sqrt(24*365) if returns_arr.std() > 0 else np.nan

    test_results.append({
        "fold": fold_idx + 1, "n_trades": n_trades,
        "win_rate_%": round(win_rate * 100, 1),
        "equity_final": round(equity_curve[-1], 3),
        "max_drawdown_%": round(max_dd * 100, 1),
        "sharpe": round(sharpe, 2),
    })
    print(f"Fold {fold_idx+1} TEST — win_rate={win_rate*100:.1f}% | equity={equity_curve[-1]:.3f} | drawdown={max_dd*100:.1f}% | Sharpe={sharpe:.2f}")

test_results_df = pd.DataFrame(test_results)
print("\n=== Résumé FINAL sur TEST (verdict honnête) ===")
print(test_results_df.to_string(index=False))

"""## 7. Visualisation — reward d'entraînement et équity, par fold"""

fig, axes = plt.subplots(len(folds), 2, figsize=(13, 4 * len(folds)))
if len(folds) == 1:
    axes = axes.reshape(1, -1)

for i in range(len(folds)):
    rewards = fold_training_rewards[i]
    smoothed = pd.Series(rewards).rolling(max(3, len(rewards)//15), min_periods=1).mean()
    axes[i, 0].plot(rewards, alpha=0.3)
    axes[i, 0].plot(smoothed, linewidth=2)
    axes[i, 0].set_title(f"Fold {i+1} — reward d'entraînement")
    axes[i, 0].grid(alpha=0.3)

    axes[i, 1].plot(fold_equity_curves[i])
    axes[i, 1].set_title(f"Fold {i+1} — équity sur validation")
    axes[i, 1].grid(alpha=0.3)

plt.tight_layout()
plt.show()

# ============================================================
# BLOC DE LOGGING — à coller en TOUTE DERNIÈRE cellule du notebook.
# Enregistre ce run dans le Google Sheet partagé (onglets "Runs" et
# "Results"), quel que soit le pipeline (RL, équation, régime...).
#
# Pré-requis dans les cellules précédentes :
#   - EXPERIMENT_CONFIG, PIPELINE_TYPE, PIPELINE_CONFIG   (cellules 4-5)
#   - log_path ajouté à l'EvalCallback                     (cellule 6, si RL)
#   - folds, df_full, results_df, test_results_df déjà en mémoire
#   - REPO_PATH défini par colab_github_sync.py (cellule 1)
#
# Limite connue : si une cellule PRÉCÉDENTE plante, ce bloc ne
# s'exécute jamais et le run raté n'est pas loggé. Pour l'instant
# c'est accepté comme compromis simple — on pourra ajouter un
# try/except autour de la cellule d'entraînement plus tard si le
# besoin de tracer aussi les crashs devient important.
# ============================================================

!pip install -q gspread google-auth

import gspread
from google.colab import auth
import google.auth
from datetime import datetime, timezone
import subprocess
import json
import os
import numpy as np

# --- À définir une seule fois, identique pour toi et ton collègue ---
GOOGLE_SHEET_URL = "https://docs.google.com/spreadsheets/d/XXXXXXXXXXXX/edit"
CRASH_REWARD_THRESHOLD = -50  # reward en dessous = run marqué "crashed_during_training"
# ----------------------------------------------------------------

RUNS_HEADER = ["run_id", "timestamp", "pipeline_type", "git_commit", "code_dirty",
               "status", "hypothesis", "experiment_config", "pipeline_config", "comment"]
RESULTS_HEADER = ["run_id", "fold", "split", "n_trades", "win_rate_%", "equity_final",
                   "max_drawdown_%", "sharpe", "buy_hold_equity", "pipeline_diagnostics"]


def _git_info(repo_path):
    def run(cmd):
        r = subprocess.run(cmd, shell=True, cwd=repo_path, capture_output=True, text=True)
        return r.stdout.strip()
    if not repo_path or not os.path.exists(repo_path):
        return "no_git", True
    commit = run("git rev-parse --short HEAD") or "unknown"
    dirty = bool(run("git status --porcelain"))
    return commit, dirty


def _buy_hold_equity(df_full, start_idx, end_idx):
    seg = df_full["close"].iloc[start_idx:end_idx]
    return round(float(seg.iloc[-1] / seg.iloc[0]), 3)


def _fold_training_diagnostics(fold_idx, crash_threshold=CRASH_REWARD_THRESHOLD):
    """Lit evaluations.npz (sauvegardé grâce à log_path dans EvalCallback)
    pour retrouver le step du meilleur checkpoint et détecter un crash
    d'entraînement. Renvoie un dict vide si non disponible (ex: pipeline
    non-RL, ou log_path pas encore ajouté)."""
    path = f"./best_model_fold{fold_idx}/evaluations.npz"
    if not os.path.exists(path):
        return {"best_checkpoint_step": None, "crashed_during_training": None}
    data = np.load(path)
    mean_rewards = data["results"].mean(axis=1)
    timesteps = data["timesteps"]
    best_step = int(timesteps[int(mean_rewards.argmax())])
    crashed = bool((mean_rewards < crash_threshold).any())
    return {"best_checkpoint_step": best_step, "crashed_during_training": crashed}


def _get_or_create_worksheet(sheet, name, header):
    try:
        ws = sheet.worksheet(name)
    except gspread.exceptions.WorksheetNotFound:
        ws = sheet.add_worksheet(title=name, rows=1000, cols=len(header))
        ws.append_row(header)
    return ws


# ------------------ Saisie humaine (les deux seules choses tapées à la main) ------------------
hypothesis = input("Hypothèse testée dans ce run (une phrase) : ").strip()
comment = input("Conclusion / commentaire (optionnel, Entrée pour passer) : ").strip()

# ------------------ Assemblage des infos du run ------------------
repo_path = globals().get("REPO_PATH", None)
commit, dirty = _git_info(repo_path)
pipeline_type = globals().get("PIPELINE_TYPE", "unspecified")
run_id = f"{datetime.now(timezone.utc).strftime('%Y%m%d-%H%M')}_{pipeline_type}"

run_row = [
    run_id,
    datetime.now(timezone.utc).isoformat(),
    pipeline_type,
    commit,
    str(dirty),
    "completed",
    hypothesis,
    json.dumps(globals().get("EXPERIMENT_CONFIG", {})),
    json.dumps(globals().get("PIPELINE_CONFIG", {})),
    comment,
]

# ------------------ Assemblage des résultats, par fold x split ------------------
result_rows = []
for fold_idx, f in enumerate(folds):
    diag = _fold_training_diagnostics(fold_idx)
    bh_equity = _buy_hold_equity(df_full, f["test"][0], f["test"][1])

    val_row = results_df[results_df["fold"] == fold_idx + 1].iloc[0]
    result_rows.append([
        run_id, fold_idx + 1, "val",
        int(val_row["n_trades"]), float(val_row["win_rate_%"]), float(val_row["equity_final"]),
        float(val_row["max_drawdown_%"]), float(val_row["sharpe_annualized"]),
        "", json.dumps(diag),
    ])

    test_row = test_results_df[test_results_df["fold"] == fold_idx + 1].iloc[0]
    result_rows.append([
        run_id, fold_idx + 1, "test",
        int(test_row["n_trades"]), float(test_row["win_rate_%"]), float(test_row["equity_final"]),
        float(test_row["max_drawdown_%"]), float(test_row["sharpe"]),
        bh_equity, json.dumps(diag),
    ])

# ------------------ Écriture Google Sheets ------------------
auth.authenticate_user()
creds, _ = google.auth.default()
gc = gspread.authorize(creds)
sh = gc.open_by_url(GOOGLE_SHEET_URL)

runs_ws = _get_or_create_worksheet(sh, "Runs", RUNS_HEADER)
results_ws = _get_or_create_worksheet(sh, "Results", RESULTS_HEADER)

runs_ws.append_row(run_row, value_input_option="USER_ENTERED")
results_ws.append_rows(result_rows, value_input_option="USER_ENTERED")

print(f"✅ Run '{run_id}' enregistré ({len(result_rows)} lignes de résultats, "
      f"commit {commit}{' [modifié, non commité]' if dirty else ''}).")

"""## 8. Notes de lecture des résultats

**Sur le Sharpe :** calculé sur la courbe d'équity (pas sur les trades individuels, trop peu nombreux pour être fiable). Un Sharpe annualisé > 3 sur un échantillon de quelques centaines/milliers de points reste à prendre avec prudence — plus probablement un signe d'échantillon encore limité qu'une vraie edge exceptionnelle (repère réel : 1-2 est déjà bon, 2-3 excellent).

**Sur le Sortino (volontairement absent ici) :** avec un SL fixe, les pertes ont presque toutes la même taille → écart-type des pertes proche de zéro → le Sortino explose numériquement (observé en V1 : valeur à 235 000 milliards). Pas une métrique fiable pour ce design de récompense, le Sharpe seul suffit.

**Sur la variance :** un seul seed par fold ici. Si tu as le temps, relance la cellule 6 avec `SEED = 456` puis `SEED = 789` (change uniquement cette ligne) pour vérifier, comme en V1, que les résultats restent cohérents d'un seed à l'autre — surtout si un fold donne un résultat qui te semble suspicieusement bon ou mauvais.

**Prochaines étapes (avec ton ami) :** passer à `N_FOLDS=4`, tester plusieurs `TP_MULT` (1.5, 2.0, 2.5, 3.0) en moyennant sur plusieurs seeds pour chacun, et tuning des hyperparamètres PPO — hors scope de ce prototype solo.
"""
```
